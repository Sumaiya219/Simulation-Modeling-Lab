```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1.Create two 5x5 matrices with random integers
A = np.random.randint(1, 10, (5, 5))
B = np.random.randint(1, 10, (5, 5))

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

# Addition
print("\nAddition (A + B):")
print(A + B)

# Subtraction
print("\nSubtraction (A - B):")
print(A - B)

# Multiplication
print("\nMultiplication (A × B):")
print(np.dot(A, B))

# Determinant
det_A = np.linalg.det(A)
print("\nDeterminant of A:")
print(det_A)

# Inverse
if det_A != 0:
    print("\nInverse of A:")
    print(np.linalg.inv(A))
else:
    print("\nA has no inverse.")

# Rank
print("\nRank of A:")
print(np.linalg.matrix_rank(A))
```

```python
# 2.a. Create a 4x4 matrix with random values
matrix = np.random.randint(1, 20, (4, 4))

print("4x4 Random Matrix:")
print(matrix)

# b & c. Display matrix using heatmap with row and column labels
plt.figure(figsize=(7, 5))

sns.heatmap(
    matrix,
    annot=True,
    fmt="d",
    cmap="viridis",
    xticklabels=["Column 1", "Column 2", "Column 3", "Column 4"],
    yticklabels=["Row 1", "Row 2", "Row 3", "Row 4"]
)

plt.title("Heatmap of 4x4 Random Matrix")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```
