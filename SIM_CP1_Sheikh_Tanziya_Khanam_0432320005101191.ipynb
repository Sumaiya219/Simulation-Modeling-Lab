```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# TASK 1: ADVANCED SCALAR & ARRAY OPERATIONS
scalar = 4.3
array = np.array([-4.5, -2.2, 0.8, 1.5, 2.7, 3.6, 5.1, 7.4])
print("\nScalar:", scalar)
print("Array:", array)

print("\n--- Arithmetic Operations ---")

print("Addition with scalar:")
print(array + scalar)

print("\nSubtraction with scalar:")
print(array - scalar)

print("\nMultiplication with scalar:")
print(array * scalar)

print("\nDivision with scalar:")
print(array / scalar)

print("\n--- Mathematical Functions ---")

safe_array = np.abs(array)

print("Square Root:")
print(np.sqrt(safe_array))

print("\nAbsolute Value:")
print(np.abs(array))

print("\nExponential:")
print(np.exp(array))

print("\nNatural Logarithm:")
print(np.log(safe_array))
print("\nLogarithm Base 10:")
print(np.log10(safe_array))

print("\n--- Trigonometric Functions ---")

print("sin():")
print(np.sin(array))

print("\ncos():")
print(np.cos(array))

print("\ntan():")
print(np.tan(array))

clipped_array = np.clip(array, -1, 1)

print("\narcsin() after clipping:")
print(np.arcsin(clipped_array))

print("\narccos() after clipping:")
print(np.arccos(clipped_array))
print("\narctan():")
print(np.arctan(array))
```

```python
# TASK 2: MATRIX MATHEMATICAL OPERATIONS
print("\n" + "=" * 60)
print("TASK 2: MATRIX MATHEMATICAL OPERATIONS")
print("=" * 60)
matrix = np.array([
    [1.2, -2.5, 3.7],
    [-4.1, 5.3, -6.2],
    [2.8, -1.6, 4.5],
    [-3.3, 6.7, -0.9]
])

print("\n4x3 Matrix:")
print(matrix)
print("\n--- Trigonometric Operations ---")

print("sin(matrix):")
print(np.sin(matrix))

print("\ncos(matrix):")
print(np.cos(matrix))

print("\ntan(matrix):")
print(np.tan(matrix))

print("\n--- Inverse Trigonometric Operations ---")

# Clip values to [-1, 1]
clipped_matrix = np.clip(matrix, -1, 1)

print("arcsin(matrix):")
print(np.arcsin(clipped_matrix))

print("\narccos(matrix):")
print(np.arccos(clipped_matrix))

print("\narctan(matrix):")
print(np.arctan(matrix))

print("\n--- Exponential & Logarithmic Operations ---")

safe_matrix = np.abs(matrix)

print("exp(matrix):")
print(np.exp(matrix))

print("\nlog(matrix) using absolute values:")
print(np.log(safe_matrix))

print("\nlog10(matrix) using absolute values:")
print(np.log10(safe_matrix))

print("\n--- Other Mathematical Operations ---")

print("Absolute Value:")
print(np.abs(matrix))

print("\nSquare Root:")
print(np.sqrt(np.abs(matrix)))

print("\nSquare:")
print(np.square(matrix))

print("\nCube Root:")
print(np.cbrt(matrix))

print("\n--- Rounding and Remainder Operations ---")

print("Remainder when divided by 3:")
print(np.mod(matrix, 3))

print("\nRound:")
print(np.round(matrix))

print("\nFloor:")
print(np.floor(matrix))

print("\nCeil:")
print(np.ceil(matrix))
```

```python
# TASK 3: MATRIX STATISTICS AND ANALYSIS
print("\n" + "=" * 60)
print("TASK 3: MATRIX STATISTICS AND ANALYSIS")
print("=" * 60)

np.random.seed(42)

stats_matrix = np.random.uniform(-10, 10, (3, 4))

print("\n3x4 Random Matrix:")
print(stats_matrix)
print("\n--- Basic Statistics ---")

maximum = np.max(stats_matrix)
minimum = np.min(stats_matrix)

max_index = np.unravel_index(
    np.argmax(stats_matrix),
    stats_matrix.shape
)

min_index = np.unravel_index(
    np.argmin(stats_matrix),
    stats_matrix.shape
)
print("Maximum value:", maximum)
print("Minimum value:", minimum)
print("Index position of maximum value:", max_index)
print("Index position of minimum value:", min_index)
print("Sum of all elements:", np.sum(stats_matrix))
print("Product of all elements:", np.prod(stats_matrix))
print("Mean:", np.mean(stats_matrix))
print("Median:", np.median(stats_matrix))
print("Variance:", np.var(stats_matrix))
print("Standard Deviation:", np.std(stats_matrix))
print("\n--- Row and Column Analysis ---")

print("Sum of each row:")
print(np.sum(stats_matrix, axis=1))

print("\nSum of each column:")
print(np.sum(stats_matrix, axis=0))

print("\nMaximum value from each row:")
print(np.max(stats_matrix, axis=1))

print("\nMinimum value from each column:")
print(np.min(stats_matrix, axis=0))
print("\n" + "=" * 60)
print("ALL TASKS COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
