```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Task 1:Scalar & Array Operations
scalar = 3.7
array = np.array([2, 5, 8, 10, 12, 15])

print("Scalar =", scalar)
print("Array =", array)

print("sin =", np.sin(scalar))
print("cos =", np.cos(scalar))
print("tan =", np.tan(scalar))

print("exp =", np.exp(scalar))
print("log =", np.log(scalar))

print("sqrt =", np.sqrt(scalar))
print("square =", np.square(scalar))
print("absolute =", np.abs(scalar))

print("round =", np.round(scalar))
print("floor =", np.floor(scalar))
print("ceil =", np.ceil(scalar))
```

```python
# Task 2: 3x2 Matrix Operations
matrix = np.array([
    [1.2, 2.5],
    [3.8, 4.6],
    [5.4, 6.9]
])

print("Matrix:")
print(matrix)

print("\nTrigonometric Functions")
print("sin:\n", np.sin(matrix))
print("cos:\n", np.cos(matrix))
print("tan:\n", np.tan(matrix))

print("\nExponential")
print(np.exp(matrix))

print("\nLogarithm")
print(np.log(matrix))

print("\nAbsolute")
print(np.abs(matrix))

print("\nSquare Root")
print(np.sqrt(matrix))

print("\nRemainder when divided by 2")
print(np.mod(matrix,2))

print("\nRound")
print(np.round(matrix))

print("\nFloor")
print(np.floor(matrix))

print("\nCeil")
print(np.ceil(matrix))
```

```python
# Task 4: Vector Functions on 2x3 Matrix
A = np.array([
    [2.1, 4.7, 1.5],
    [8.2, 6.3, 3.9]
])

print("Matrix:")
print(A)

print("\nMaximum Value:", np.max(A))
print("Index of Maximum:", np.unravel_index(np.argmax(A), A.shape))

print("\nMinimum Value:", np.min(A))
print("Index of Minimum:", np.unravel_index(np.argmin(A), A.shape))

print("\nVector Length:")
print(np.linalg.norm(A))

print("\nRow-wise Descending Sort")
print(np.sort(A, axis=1)[:,::-1])

print("\nEntire Matrix Descending")
print(np.sort(A.flatten())[::-1])

print("\nSum =", np.sum(A))
print("Product =", np.prod(A))

print("\nMedian =", np.median(A))
print("Mean =", np.mean(A))
print("Standard Deviation =", np.std(A))
```

```python
# Heatmap
import matplotlib.pyplot as plt

plt.imshow(A, cmap="viridis")
plt.colorbar()
plt.title("Matrix Heatmap")
plt.show()
```
